```python
# STep 8: Inspect sales columns

sales_columns = [
    column for column in df.columns
    if any(keyword in column.lower()
           for keyword in ["sales", "revenue", "price", "amount", "quantity"])
]

print("Potential sales-related columns:")
print(sales_columns)
```

```python
# STep 9: Check customer/order information

customer_columns = [
    column for column in df.columns
    if any(keyword in column.lower()
           for keyword in ["customer", "client", "user"])
]

print("Potential customer-related columns:")
print(customer_columns)
```

```python
order_columns = [
    column for column in df.columns
    if any(keyword in column.lower()
           for keyword in ["order", "transaction", "invoice"])
]

print("Potential order-related columns:")
print(order_columns)
```

# Step 10 :Check whether CAC can actually be calculated

CAC means:

Customer Acquisition Cost

The general calculation is:(CAC = Customer Acquisition Costs / Number of New Customers)

The dataset needs information about acquisition/marketing cost.

```python
cac_columns = [
    column for column in df.columns
    if any(keyword in column.lower()
           for keyword in [
               "acquisition",
               "marketing",
               "campaign",
               "advertising",
               "ad_spend",
               "marketing_cost",
               "acquisition_cost"
           ])
]

print("Potential CAC-related columns:")
print(cac_columns)
```

```python
# Step 11:Check whether churn can actually be calculated

churn_columns = [
    column for column in df.columns
    if any(keyword in column.lower()
           for keyword in [
               "churn",
               "status",
               "active",
               "inactive",
               "retention"
           ])
]

print("Potential churn-related columns:")
print(churn_columns)
```

```python
# Step 12: Create an initial KPI calculation

revenue = df["net_sales"].sum()

print("Total Revenue:", revenue)
```

```python
# STep 13: Calculate number of transactions

"order_id" in df.columns
```

```python
total_orders = df["order_id"].nunique()

print("Total Orders:", total_orders)
```

```python
# STep 14: Calculate AOV

#Average Order Value is:AOV = Total Revenue / Total Orders
#If order_id exists:

aov = revenue / total_orders

print("Average Order Value:", round(aov, 2))
```

```python
#Step 15: Create a KPI summary

kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Revenue",
        "Total Orders",
        "Average Order Value"
    ],
    
    "Value": [
        revenue,
        total_orders,
        aov
    ]
})

kpi_summary
```

```python
# Step 16: Save the KPI summary

kpi_summary.to_csv(r"C:\Users\HP\OneDrive\Desktop\Retail_Dashboard_Project\output\kpi_summary.csv",index=False)

print("KPI summary saved successfully.")
```

```python
# Step 17: Create monthly revenue data

monthly_revenue = (
    df.groupby(["year", "month"])["net_sales"]
      .sum()
      .reset_index()
)

monthly_revenue
```

```python
# Create a period

monthly_revenue["period"] = (
    monthly_revenue["year"].astype(str)
    + "-"
    + monthly_revenue["month"].astype(str).str.zfill(2)
)

monthly_revenue
```

```python
# Step 18: Save monthly revenue

monthly_revenue.to_csv(r"C:\Users\HP\OneDrive\Desktop\Retail_Dashboard_Project\output\monthly_revenue.csv",index=False)

print("Monthly revenue data saved successfully.")
```

```python
# Step 19: Create category-level data

category_revenue = (
    df.groupby("category")["net_sales"]
      .sum()
      .reset_index()
      .sort_values("net_sales", ascending=False)
)

category_revenue
```

```python
category_revenue.to_csv(r"C:\Users\HP\OneDrive\Desktop\Retail_Dashboard_Project\output\category_revenue.csv",index=False)
```

```python
# Step 20: Create customer-segment data

segment_revenue = (
    df.groupby("customer_segment")["net_sales"]
      .sum()
      .reset_index()
      .sort_values("net_sales", ascending=False)
)

segment_revenue
```

```python
segment_revenue.to_csv(r"C:\Users\HP\OneDrive\Desktop\Retail_Dashboard_Project\output\segment_revenue.csv",index=False)
```

```python
# Step 21: Create geographic data

city_revenue = (
    df.groupby("city")["net_sales"]
      .sum()
      .reset_index()
      .sort_values("net_sales", ascending=False)
)

city_revenue.head(20)
```

```python
city_revenue.to_csv(r"C:\Users\HP\OneDrive\Desktop\Retail_Dashboard_Project\output\city_revenue.csv",index=False)
```

```python
# STep 22:Create dashboard-ready dataset

df.to_csv(r"C:\Users\HP\OneDrive\Desktop\Retail_Dashboard_Project\output\dashboard_data.csv",index=False)

print("Dashboard dataset saved successfully.")
```
